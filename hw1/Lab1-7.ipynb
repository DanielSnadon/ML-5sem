````python
import pandas as pd
````

````python
data = pd.read_csv("data.csv", keep_default_na=False)
data.head()
````

1. Какой средний вес людей представленных в наборе данных?

````python
averageWeight = data["Weight (kg)"].mean()
print("Средний вес:", averageWeight, "кг")
````

2. Сбалансировано ли соотношение мужчин и женщин в наборе данных?

````python
genderCounts = data["Gender"].value_counts()
genderPercent = data["Gender"].value_counts(normalize=True) * 100
print("Количество людей разного пола:")
print(genderCounts)
print("\nДоля людей разного пола:")
print(genderPercent)

print("\nОтвет: соотношение можно считать примерно сбалансированным.")
````

3. Правда ли, что люди с высоким уровнем стресса имеют более высокое кровяное давление, чем те, кто не подвержен стрессу

````python
pressure = data["Blood Pressure (s/d)"].str.split("/", expand=True)
data["Systolic"] = pd.to_numeric(pressure[0], errors="coerce")
data["Diastolic"] = pd.to_numeric(pressure[1], errors="coerce")

lowLimit = data["Stress Levels"].quantile(0.25)
highLimit = data["Stress Levels"].quantile(0.75)

lowStress = data[data["Stress Levels"] <= lowLimit]
highStress = data[data["Stress Levels"] >= highLimit]

pressureColumns = ["Systolic", "Diastolic"]

print("\nЛюдей с низким стрессом:", len(lowStress))
print("Среднее давление:")
print(lowStress[pressureColumns].mean())

print("\nЛюдей с высоким стрессом:", len(highStress))
print("Среднее давление:")
print(highStress[pressureColumns].mean())

print("\nОтвет: при высоком стрессе давление выше, но не значительно.")
````

4. Правда ли, что уровень холестерина у мужчин и женщин отличается?

````python
cholesterol = data.groupby("Gender")["Cholesterol Level (mg/dL)"].mean()
print("Средний уровень холестерина:")
print(cholesterol)

print("\nОтвет: уровень холестерина у мужчин и женщин отличается, но не значительно.")
````

5. Правда ли, что с возрастом женщины имеют свойство оценивать свои когнитивные способности хуже, чем в юности (порог юности - 30 лет, женщина в возрасте - 60 лет+)

````python
women = data[data["Gender"] == "Female"]
youngWomen = women[women["Age (years)"] < 30]
olderWomen = women[women["Age (years)"] >= 60]
youngScore = youngWomen["Cognitive Function"].mean()
olderScore = olderWomen["Cognitive Function"].mean()

print("Young женщины:")
print("Количество:", len(youngWomen))
print("Средняя оценка:", youngScore)

print("\nOld женщины:")
print("Количество:", len(olderWomen))
print("Средняя оценка:", olderScore)

print("\nОтвет: действительно, оценка группы женщин старше 60 лет значительно ниже.")
````

6. Влияет ли физическая активность на вес в возрасте более 50 лет?

````python
olderPeople = data[data["Age (years)"] > 50]
weight = olderPeople.groupby("Physical Activity Level")["Weight (kg)"].agg(["mean", "count"])

print("Вес людей старше 50 лет по уровню активности:")
print(weight)

print("\nОтвет: да, физическая активность немного влияет на вес в возрасте более 50 лет.")
````

7. Сравните минимальный вес мужчин с высоким и низким заработком

````python
men = data[data["Gender"] == "Male"]
weightByIncome = men.groupby("Income Level")["Weight (kg)"].min()

print("Низкий доход:", weightByIncome["Low"], "кг")
print("Высокий доход:", weightByIncome["High"], "кг")

print("\nОтвет: минимальный вес мужчин с высоким заработком меньше, чем у мужчин с низким заработком.")
````

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python
bmi30 = data["BMI"].quantile(0.30)
bmi60 = data["BMI"].quantile(0.60)
bmi75 = data["BMI"].quantile(0.75)

middleBmi = data[(data["BMI"] >= bmi30) & (data["BMI"] <= bmi60)]
highBmi = data[data["BMI"] > bmi75]

middleDiseases = middleBmi["Chronic Diseases"] != "None"
highDiseases = highBmi["Chronic Diseases"] != "None"

print("ИМТ от 30-го до 60-го перцентиля:")
print("Людей в группе:", len(middleBmi))
print("Людей с заболеваниями:", middleDiseases.sum())
print("Доля с заболеваниями:", middleDiseases.mean() * 100, "%")

print("\nИМТ выше 75-го перцентиля:")
print("Людей в группе:", len(highBmi))
print("Людей с заболеваниями:", highDiseases.sum())
print("Доля с заболеваниями:", highDiseases.mean() * 100, "%")

print("\nОтвет: да, у людей с ИМТ больше 75-го перцентиля хронические заболевания встречаются чаще.")
````

9. Правда ли, что люди с высоким уровнем дохода имеют больший уровень физической активности, чем люди с низким доходом

````python
incomeGroups = data[data["Income Level"].isin(["Low", "High"])]
activityPercent = pd.crosstab(incomeGroups["Income Level"], incomeGroups["Physical Activity Level"], normalize="index") * 100
print(activityPercent)

print("\nОтвет: нет, люди с высоким уровнем дохода не имеют больший уровень физической активности.")
````

10. Какая доля мужчин с высокой частотой пребывания на солнце имеет высокий заработок?

````python
men = data[data["Gender"] == "Male"]
sunLimit = men["Sun Exposure"].quantile(0.75)
sunnyMen = men[men["Sun Exposure"] >= sunLimit]

print("Мужчин с высоким пребыванием на солнце:", len(sunnyMen))

richMen = sunnyMen[sunnyMen["Income Level"] == "High"]

print("Из них с высоким доходом:", len(richMen))

share = len(richMen) / len(sunnyMen) * 100

print("Доля мужчин с высоким доходом:", share, "%")
````
